Problem 3

```python
#Creating dictionary  to contain the coordinates in Bohrs

atoms = [{'symbol': 'O', 'x': 0.00000, 'y': 0.0000, 'z': 0.0000},
                     {'symbol': 'H', 'x': 1.80941647, 'y': 0.000, 'z': 0.000},
                     {'symbol': 'H', 'x': -0.45334744, 'y': 1.75170319, 'z': 0.000}
]
```

```python
# Cinversion factor from Bohr to Angstrom
conv_factor = 0.529177249
```

```python
# Computing the conversion including if statement and for loop

units = 'bohr'
atoms_ang = []
if  units.lower() == 'bohr':
    for atom in atoms:
        atoms_ang.append({ 'symbol': atom['symbol'],

                                        'x': atom['x']*conv_factor,
                                        'y': atom['y']*conv_factor,
                                        'z': atom['z']*conv_factor
        })
else:
    atoms_ang = atoms
```

```python
#  Creating an XYZ file using the coordinates in Angstrom
with open("water.xyz", "w") as f:
    f.write(f"3\n")
    f.write(f"Water molecule (units converted from {units} to angstrom)\n")
    for atom in atoms_ang  :
             f.write(f"{atom['symbol']:<2s} {atom['x']: .6f} {atom['y']: .6f} {atom['z']: .6f}\n")
          
```

```python
atoms_ang
```

Problem 4a

```python
# Oxygen atom's Coordinates(x,y,z) in Bohr)
x_O = atoms[0]['x'] 
y_O = atoms[0]['y']
z_O = atoms[0]['z']

# First Hydrogen atom's Coordinates(x,y,z) in Bohr)
x_H1 = atoms[1]['x'] 
y_H1 = atoms[1]['y']
z_H1 = atoms[1]['z']

# Second Hydrogen atom's Coordinates(x,y,z) in Bohr)
x_H2 = atoms[1]['x'] 
y_H2 = atoms[1]['y']
z_H2 = atoms[1]['z']
```

```python
x_H2
```

```python
# Calculating the distance between Oxygen atom(O) and the first Hydrogen atom (H1)as d_OH1

d_OH1 = np.sqrt(((x_O-x_H1)**2)+((y_O-y_H1)**2)+((z_O-z_H1)**2))
```

```python
print(float(d_OH1),'bohrs')
```

```python
# Calculating the distance between Oxygen atom(O) and the second Hydrogen atom (H2)as d_OH2

d_OH2 = np.sqrt(((x_O-x_H2)**2)+((y_O-y_H2)**2)+((z_O-z_H2)**2))
```

```python
print(float(d_OH2),'bohrs')
```
